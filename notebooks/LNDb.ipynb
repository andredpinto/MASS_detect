```python
import pandas as pd
all_nods = pd.read_csv('../data/LNDb/allNods.csv')
gt_nods = pd.read_csv('../data/LNDb/train/trainset_csv/trainNodules_gt.csv')

print(f"The table all_nods has the following columns:\n{', '.join(all_nods.columns)}\nand has {all_nods.shape[0]} rows and {all_nods.shape[1]} columns in total.")
```

```python
all_nods = all_nods[all_nods['Where'].str.contains("RadAnnotation", na=False)]
all_nods= all_nods.drop(columns=['TextInstanceID', 'TextQuestion', 'Diam_Text', 'Caract_Text', 'Where'], axis=1)

print(f"The table all_nods has the following columns:\n{', '.join(all_nods.columns)}\nand has {all_nods.shape[0]} rows and {all_nods.shape[1]} columns in total.")
```

```python
only_in_all_nods = set(all_nods['RadID']) - set(gt_nods['RadID'])
print(only_in_all_nods)

print(all_nods[all_nods['RadID'].isin(only_in_all_nods)].shape)

print(gt_nods["RadID"].value_counts())
```

```python
char_train= pd.read_csv('../data/LNDb/train/trainset_csv/chars_trainNodules.csv')
train = pd.read_csv('../data/LNDb/train/trainset_csv/trainNodules.csv')
train = train.astype('float64')
# 2. Identify common join keys between the files
join_keys = ['LNDbID', 'RadID', 'FindingID']

# 'outer' keeps all entries, 'inner' keeps only nodules present in both.
combined_df = pd.merge(char_train, train, on=join_keys, how="outer")

# View output
print(f"Combined Shape: {combined_df.shape}")
print(combined_df['RadID'].value_counts())
```
